# Import Required Libraries

```python
import pandas as pd
import os
from pathlib import Path
```

# Configuration

```python
# Update this path to your division-wise roll list Excel file
excel_file_path = "DivisionWiseRollList201723July.xlsx"  # Change this to your actual file path
sheet_name = "Table 5"
output_csv = "input_people_csv_v2.csv"
college_name = "Indian Institute of Technology, Bombay"
```

# Load Excel Sheet Data

```python
# Check if file exists
if not Path(excel_file_path).exists():
    print(f"❌ File not found: {excel_file_path}")
    print(f"Please provide the correct path to your Excel file.")
else:
    # Read the sheet, skip first row, no header
    df = pd.read_excel(excel_file_path, sheet_name=sheet_name, skiprows=1, header=None)
    print(f"✓ Loaded sheet '{sheet_name}' from {excel_file_path}")
    print(f"Total rows: {len(df)}")
    print(f"\nFirst few rows:")
    print(df.head())
```

# Extract Roll Numbers and Names

```python
# Column indices: 2 and 3 (0-indexed: 1 and 2)
# Extract roll number (column 2, index 1) and name (column 3, index 2)
roll_numbers = df.iloc[:, 1]  # Column 2 (0-indexed as 1)
names = df.iloc[:, 2]         # Column 3 (0-indexed as 2)

print(f"✓ Extracted {len(roll_numbers)} roll numbers and names")
print(f"\nSample data:")
for i in range(min(5, len(roll_numbers))):
    print(f"  Roll: {roll_numbers.iloc[i]}, Name: {names.iloc[i]}")
```

# Create People DataFrame

```python
# Create a new dataframe with roll_number, name, and college
people_df = pd.DataFrame({
    'roll_number': roll_numbers.values,
    'name': names.values,
    'college': college_name
})

print(f"✓ Created people dataframe with {len(people_df)} records")
print(f"\nDataframe columns: {list(people_df.columns)}")
print(f"\nFirst few rows:")
print(people_df.head())
print(f"\nDataframe shape: {people_df.shape}")
```

# Export to CSV

```python
# Export to CSV without index
people_df.to_csv(output_csv, index=False, encoding='utf-8')
print(f"✅ Successfully saved to {output_csv}")
print(f"\nFile location: {Path(output_csv).absolute()}")
print(f"Total records exported: {len(people_df)}")
```

# Verification

```python
# Verify the output file
if Path(output_csv).exists():
    print(f"✓ File exists: {output_csv}")
    # Read and display the first few rows
    verify_df = pd.read_csv(output_csv)
    print(f"\nVerification - First 5 rows of exported CSV:")
    print(verify_df.head())
    print(f"\nTotal rows in CSV: {len(verify_df)}")
    print(f"\nDataframe info:")
    print(verify_df.info())
else:
    print(f"❌ File not found: {output_csv}")
```
